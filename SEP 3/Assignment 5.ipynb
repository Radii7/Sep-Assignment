```python
df['Age_mean_imputed'] = df['Age'].fillna(df['Age'].mean())
df['Age_median_imputed'] = df['Age'].fillna(df['Age'].median())
df['Cabin_mode_imputed'] = df['Cabin'].fillna(df['Cabin'].mode()[0])
df[['Age','Age_mean_imputed','Age_median_imputed','Cabin','Cabin_mode_imputed']]
df
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/titanic uncleaned.csv")
cols=['Age','SibSp','Parch','Fare']
plt.figure(figsize=(12,8))
for i,col in enumerate(cols,1):
  plt.subplot(2,2,i)
  sns.boxplot(y=df[col])
  plt.title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()
```

```python
df_clean=df.copy()
for col in cols:
  Q1 = df_clean[col].quantile(0.25)
  Q3 = df_clean[col].quantile(0.75)
  IQR = Q3 - Q1
  lower_limit = Q1 - 1.5 * IQR
  upper_limit = Q3 + 1.5 * IQR
  df_clean = df_clean[
      (df_clean[col] >= lower_limit) &
       (df_clean[col] <= upper_limit)
  ]
print("Orginal shape:",df.shape)
print("After outlier removal:",df_clean.shape)
```

```python
plt.figure(figsize=(12,8))

for i,col in enumerate(cols,1):
  plt.subplot(2,2,i)
  sns.boxplot(y=df_clean[col])
  plt.title(f'{col} - After Outlier Removal')

plt.tight_layout()
plt.show()
```

```python
from numpy import ones_like
import pandas as pd
from sklearn.preprocessing import LabelEncoder
encoded_data = df.copy()
label_encoder = LabelEncoder()
encoded_data['Sex_Label'] = label_encoder.fit_transform(encoded_data['Sex'])

one_hot_encoded_data = pd.get_dummies(encoded_data['Embarked'],prefix='Embarked')
encoded_data = pd.concat([encoded_data,one_hot_encoded_data],axis=1)

pclass_mapping = {1:'First',2:'Second',3:'Third'}
encoded_data['Pclass_Category'] = encoded_data['Pclass'].map(pclass_mapping)
print(encoded_data[['Sex','Sex_Label','Embarked_Q','Embarked_S','Embarked_C','Pclass','Pclass_Category']])
```

```python
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("After removing duplicates:", df.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Age','Fare']] = scaler.fit_transform(
    df[['Age','Fare']]
)

print(df[['Age','Fare']].head())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['Age','Fare']] = scaler.fit_transform(
    df[['Age','Fare']]
)

print(df[['Age','Fare']].head())
```
